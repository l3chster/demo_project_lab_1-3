```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()

SILVER_TABLE = "dbr_dev.weather_silver.unified_weather"
GOLD_PROFILES_TABLE = "dbr_dev.weather_gold.gold_city_profiles"

print("Starting Reconciliation Check: Silver vs Gold...")

silver_count = spark.read.table(SILVER_TABLE).count()
```

```python
gold_df = spark.read.table(GOLD_PROFILES_TABLE)
gold_count = gold_df.selectExpr("sum(total_measurements_recorded)").collect()[0][0]

# Handling nulls in case of an empty database
if gold_count is None:
    gold_count = 0

print(f"Total Rows in Silver: {silver_count}")
print(f"Sum of measurements in Gold City Profiles: {gold_count}")
```

```python
assert silver_count == gold_count, f"Reconciliation Failed! Data loss detected. Silver: {silver_count}, Gold: {gold_count}"

print("SUCCESS: Reconciliation check passed! No silent data loss detected.")
```
