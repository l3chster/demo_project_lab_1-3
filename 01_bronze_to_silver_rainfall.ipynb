```python
%sql
CREATE SCHEMA IF NOT EXISTS dbr_dev.weather_silver;
```

```python
%sql
CREATE OR REPLACE TABLE dbr_dev.weather_silver.unified_rainfall
LOCATION 'abfss://dataweather@dlspl21databricks.dfs.core.windows.net/silver/tables/unified_rainfall'
AS
SELECT 
    city,
    CAST(event_timestamp AS TIMESTAMP) AS event_time,
    rain_sum,
    showers_sum,
    snowfall_sum,    
    source_filename,
    ingestion_timestamp AS bronze_ingestion_time,
    current_timestamp() AS silver_transformation_time
FROM dbr_dev.weather_bronze.ly_rainfall_data
WHERE _rescued_data IS NULL;
```
