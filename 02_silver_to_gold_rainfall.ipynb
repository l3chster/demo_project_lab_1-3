```python
%sql
CREATE SCHEMA IF NOT EXISTS dbr_dev.weather_gold;
```

```python
%sql
CREATE OR REPLACE TABLE dbr_dev.weather_gold.gold_ly_agg_measurements
LOCATION 'abfss://dataweather@dlspl21databricks.dfs.core.windows.net/gold/tables/gold_ly_agg_measurements'
AS
SELECT
    city,    
    ROUND(SUM(rain_sum),2) AS total_rain,    
    ROUND(SUM(snowfall_sum),2) AS total_snowfall   
FROM dbr_dev.weather_silver.unified_rainfall
GROUP BY city;
```
