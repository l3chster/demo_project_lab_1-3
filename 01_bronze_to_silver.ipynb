```python
%sql
CREATE SCHEMA IF NOT EXISTS dbr_dev.weather_silver;
```

```python
%sql
CREATE OR REPLACE TABLE dbr_dev.weather_silver.unified_weather
LOCATION 'abfss://dataweather@dlspl21databricks.dfs.core.windows.net/silver/tables/unified_weather'
AS
SELECT 
    city,
    CAST(event_timestamp AS TIMESTAMP) AS event_time,
    temperature,
    wind_direction,
    wind_speed,
    air_quality,
    pm10,
    pm2_5,
    source_filename,
    ingestion_timestamp AS bronze_ingestion_time,
    current_timestamp() AS silver_transformation_time
FROM dbr_dev.weather_bronze.weather_streaming_data
WHERE _rescued_data IS NULL;
```
